```python
import json

def load_data(filename):
    with open(filename, "r") as file:
        return json.load(file)

def find_people_you_may_known(user_id, data):
    user_friends = {}
    for user in data["users"]:
        user_friends[user["id"]] = set(user["friends"])

    if user_id not in user["friends"]:
        return "---User Dosen't Exits---"



# Load the data..

data = load_data("massive_data.json")
user_id = 1
recc = find_people_you_may_known(user_id, data)
print(recc)

```
