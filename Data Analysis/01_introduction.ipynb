```python
import json
```

```python
# lets write a function  to load a data...

def load_data(filename):
    with open(filename, "r") as file :
        data = json.load(file)
    return data
```

```python
# Display the user and there connections...

def display_users(data):
    print("\n # User and there connections:\n")
    for user in data["users"]:
        print(f"ID : {user['id']} - {user['name']} - is Friends with - {user['friends']} and the liked pages are {user['liked_pages']}...")
    
    print("\n # Pages Information:\n ")
    for page in data["pages"]:
        print(f"{page['id']} : {page['name']}")

display_users(data)
```
