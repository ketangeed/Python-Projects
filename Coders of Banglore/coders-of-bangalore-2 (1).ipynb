```python
with open("initialdata.txt", "r", encoding = 'utf-8') as f:
    data = f.read()
```

```python
data
```

```python
chunks = data.split("\n\n")
```

```python
chunks[1]
```

```python
def parse_chunk(chunk):
    chunk = chunk.strip()
    sep_chunk = chunk.split("\n")
    user_name = sep_chunk[0]
    no_of_posts = sep_chunk[1]
    no_of_followers = sep_chunk[2]
    no_of_following = sep_chunk[3]
    name = sep_chunk[4]
    type_of_page = sep_chunk[5]
    bio = "\n".join(sep_chunk[6:])
    return  { "username" : user_name,"no_of_posts" : no_of_posts,"no_of_followers" : no_of_followers,"no_of_following" : no_of_following, 
              "name" : name, "type_of_page" : type_of_page,"bio" : bio}

parse_chunk(chunks[1])
```

```python
all_chunks = []
for chunk in chunks:
    parsed_chunk = parse_chunk(chunk)
    all_chunks.append(parsed_chunk)
print(all_chunks)
```
