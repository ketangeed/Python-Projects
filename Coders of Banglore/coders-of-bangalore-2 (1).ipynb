# Data Collection
You hire Vijyalaxmi Iyer and Sam Chandra who are from Hebbal and HSR Layout respectively

HSR layout and Hebbal are far away so you set up a meeting at Rameshwaram cafe, Indiranagar

```python
with open("finaldata.txt", "r", encoding='utf-8') as f:
        chunks = f.read()
```

```python
chunks = chunks.split("\n\n")
chunks = [c for c in chunks if len(c)>3]
```

```python
def parse_chunk(chunk):
    chunk = chunk.strip()
    sep_chunk = chunk.split('\n')
    user_name = sep_chunk[0]
    
    no_of_posts = int(sep_chunk[1].split(' post')[0]. replace("," , ""))

    
    no_of_followers = float(sep_chunk[2].split(' follower')[0].replace("," , "").replace("K", "").replace("M", ""))
    
    if ("K" in sep_chunk[2]):
        no_of_followers = int(no_of_followers * 1000)
    elif("M" in sep_chunk[2]):
        no_of_followers = int(no_of_followers * 1000000)
    else:
        no_of_followers = int(no_of_followers)

        
    no_of_following = int(sep_chunk[3].split(' followin')[0].replace(",", "").replace("K","").replace("M",""))
    if ("K" in sep_chunk[3]):
        no_of_following = int(no_of_following * 1000)
    elif("M" in sep_chunk[3]):
        no_of_following = int(no_of_following * 1000000)
    else:
        no_of_following = int(no_of_following)

        
    name = sep_chunk[4]

    if len(sep_chunk) > 5 and len(sep_chunk[5]) > 5:
        type_of_page = sep_chunk[5]
        bio = "\n".join(sep_chunk[6:])
    else:
        type_of_page = "unknown"
        bio = "unknown"
    
    return {"user_name" : user_name, "no_of_posts" : no_of_posts, "no_of_followers" : no_of_followers, "no_of_following" : no_of_following, 
            "name" : name, "type_of_page" : type_of_page, "bio" : bio}
```

```python
all_chunks = []
for chunk in chunks:
    parsed_chunk = parse_chunk(chunk)
    all_chunks.append(parsed_chunk)
```

```python
import json
s = json.dumps(all_chunks, indent = 4)

with open("data.json", "w") as f:
    f.write(s)
```

### who has maximum posts

```python
max_posts = 0
for chunk in all_chunks:
    if(max_posts < chunk['no_of_posts']):
        max_posts = chunk['no_of_posts']
        chunk_with_max_post = chunk
print(f"name : {chunk_with_max_post['user_name']},  posts :  {max_posts}")
```

#### who has maximum followers

```python
max_followers = 0
for chunk in all_chunks:
    if(max_followers < chunk['no_of_followers']):
        max_followers = chunk['no_of_followers']
        chunk_with_max_followers = chunk
print(f"max_follower : {(chunk_with_max_followers['user_name'])}, followers : {max_followers}")
```

#### who has maximum following

```python
max_following = 0
for chunk in all_chunks:
    if(max_following < chunk['no_of_following']):
        max_following = chunk['no_of_following']
        chunk_with_max_following = chunk
print(f"max_following : {chunk_with_max_following['user_name']},  following : {max_following}")
```

### categories

```python
category = set()
for chunk in all_chunks:
    category.add(chunk['type_of_page'])
print(f"{category}\nlen_of_category:  {len(category)}")
```
